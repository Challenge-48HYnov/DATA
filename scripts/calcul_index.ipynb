```python
import pandas as pd

df_final = pd.read_csv(r"C:/Users/PC/DATA/final/dataset_final.csv", sep=",")
```

```python
def atmo_pm25(x):
    if x <= 10: return 1
    elif x <= 20: return 2
    elif x <= 25: return 3
    elif x <= 50: return 4
    elif x <= 75: return 5
    else: return 6

def atmo_pm10(x):
    if x <= 20: return 1
    elif x <= 40: return 2
    elif x <= 50: return 3
    elif x <= 100: return 4
    elif x <= 150: return 5
    else: return 6

def atmo_no2(x):
    if x <= 40: return 1
    elif x <= 90: return 2
    elif x <= 120: return 3
    elif x <= 230: return 4
    elif x <= 340: return 5
    else: return 6
```

```python
df_final["atmo_pm25"] = df_final["PM2.5"].apply(atmo_pm25)
df_final["atmo_pm10"] = df_final["PM10"].apply(atmo_pm10)
df_final["atmo_no2"] = df_final["NO2"].apply(atmo_no2)

df_final["ATMO"] = df_final[["atmo_pm25", "atmo_pm10", "atmo_no2"]].max(axis=1)
```

```python
def atmo_label(x):
    return {
        1: "Bon",
        2: "Moyen",
        3: "Dégradé",
        4: "Mauvais",
        5: "Très mauvais",
        6: "Extrêmement mauvais"
    }[x]

df_final["ATMO_label"] = df_final["ATMO"].apply(atmo_label)
```
