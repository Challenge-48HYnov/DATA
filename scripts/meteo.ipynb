# Nettoyage du dataset meteo

### Chargement du fichier

```python
import pandas as pd

df = pd.read_csv("C:/Users/PC/DATA/raw/synop_2026.csv", sep=";")
df
```

### Sélection des colonnes utiles

```python
columns = [
    "validity_time",
    "name",
    "lat",
    "lon",
    "t",
    "u",
    "ff",
    "rr1"
]

df = df[columns]
```

### Renommer les colonnes pour une meilleures lisibilités et compréhension

```python
df = df.rename(columns={
    "validity_time": "date",
    "name": "station",
    "t": "temperature",
    "u": "Humidité",
    "ff": "wind",
    "rr1": "rain"
})
```

### Conversion de date en datetime

```python
df["date"] = pd.to_datetime(df["date"])
```

### Gestion des valeurs manquantes

```python
df = df.dropna()
```

### Sauvegarde du dataset propre

```python
df.to_csv("meteo_clean.csv", index=False)
```

##### Après vérification de  la cohérence statistique des données,  des anomalies ont été retrouvés: températures en Kelvin au lieu d'être en °C, lat et lon nous montre que les données vont au delà de la france et étant donnée que les données de pollution sont uniquement centré en france, on va centré ceux de météo sur France aussi. On  se rend compte aussi que la valeur minimal de rain est négatif, ce qui est physiquement impossible.

##### Alors, on va corriger ça

```python
# 1. Température en °C
df["temperature"] = df["temperature"] - 273.15

# 2. Filtre géographique (France)
df = df[
    (df["lat"] >= 41) & (df["lat"] <= 51) &
    (df["lon"] >= -5) & (df["lon"] <= 9)
]

# 3. Corriger pluie
df["rain"] = df["rain"].clip(lower=0)
```
